```python
import numpy as np
from scipy.stats import chi2_contingency
data = np.array([
    [10, 5],
    [5, 10]
])

chi2, p_value, dof, expected = chi2_contingency(
    data,
    correction=False
)

alpha = 0.05
critical_value = 3.841

print("Observed Frequencies:")
print(data)

print("\nExpected Frequencies:")
print(expected)

print("\nChi-square value:", chi2)
print("Degree of Freedom:", dof)

print("Critical Chi-square value:", critical_value)

if chi2 > critical_value:
    print("\nResult: Reject the Null Hypothesis")
    print("Sports is a significant feature for predicting protein powder purchase.")
else:
    print("\nResult: Fail to Reject the Null Hypothesis")
    print("Sports is NOT a significant feature for predicting protein powder purchase.")
```
