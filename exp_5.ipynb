```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([0, 0, 0, 1, 1], dtype=float)

w = 0.0
b = 0.0

alpha = 0.1
iteration = 10

print("initial weight(w):", w)
print("initial bias(b):", b)
print("learning Rate(alpha):", alpha)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def predict(x, w, b):
    z = w * x + b
    return sigmoid(z)

def compute_cost(y, y_pred):
    return -np.mean(
        y * np.log(y_pred + 1e-8) +
        (1 - y) * np.log(1 - y_pred + 1e-8)
    )

y_pred = predict(x, w, b)
print("predicted probability:", y_pred)
cost = compute_cost(y, y_pred)
print("cost:", cost)

cost_history = []
for i in range(iteration):

    y_pred = predict(x, w, b)

    dw = np.sum((y_pred - y) * x) / len(y)
    db = np.sum(y_pred - y) / len(y)

    w = w - alpha * dw
    b = b - alpha * db

    y_pred = predict(x, w, b)

    cost = compute_cost(y, y_pred)

    cost_history.append(cost)

    print("\nIteration:", i + 1)
    print("Weight(w):", w)
    print("Bias(b):", b)
    print("Cost:", cost)

y_pred = predict(x, w, b)

print("\nFinal predicted probabilities:", y_pred)
print("Final weight(w):", w)
print("Final bias(b):", b)

y_class = (y_pred >= 0.5).astype(int)

print("Final predicted classes:", y_class)
print("Actual values:", y)
```

```python
x_line = np.linspace(1, 5, 100)
y_line = predict(x_line, w, b)

plt.scatter(x, y, color="blue", label="Training Data")
plt.plot(x_line, y_line, color="red", label="Logistic Curve")

plt.xlabel("X")
plt.ylabel("Probability")
plt.title("Logistic Regression")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(
    range(1, iteration + 1),
    cost_history,
    marker='o'
)

plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost vs Iteration")
plt.grid(True)
plt.show()
```
