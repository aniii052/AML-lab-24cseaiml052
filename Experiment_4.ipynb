```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3], dtype=float)
y = np.array([2, 3, 4], dtype=float)

w = 0.0
b = 0.0

alpha = 0.05
iteration = 2

print("initial weight(w):", w)
print("initial bias(b):", b)
print("learning Rate(alpha):", alpha)


def predict(x, w, b):
    return w * x + b


def compute_cost(y, y_pred):
    return np.sum((y_pred - y) ** 2) / (2 * len(y))


y_pred = predict(x, w, b)

print("predicted value:", y_pred)

cost = compute_cost(y, y_pred)
print("cost:", cost)

for i in range(iteration):

    y_pred = predict(x, w, b)

    dw = np.sum((y_pred - y) * x) / len(y)
    db = np.sum(y_pred - y) / len(y)

    w = w - alpha * dw
    b = b - alpha * db
    y_pred = predict(x, w, b)
    cost = compute_cost(y, y_pred)

    print("\nIteration:", i + 1)
    print("Weight(w):", w)
    print("Bias(b):", b)
    print("Cost:", cost)
y_pred = predict(x, w, b)

print("\nFinal predicted values:", y_pred)
print("Final weight(w):", w)
print("Final bias(b):", b)
plt.scatter(x, y, color="blue", label="Training Data")

plt.plot(x, y_pred, color="red", label="Regression Line")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression using Gradient Descent")

plt.legend()
plt.grid(True)

plt.show()
```

```python
cost_history = []

for i in range(iteration):
    y_pred = predict(x, w, b)

    dw = np.sum((y_pred - y) * x) / len(y)
    db = np.sum(y_pred - y) / len(y)

    w = w - alpha * dw
    b = b - alpha * db

    y_pred = predict(x, w, b)
    cost = compute_cost(y, y_pred)

    cost_history.append(cost)

plt.plot(range(1, iteration + 1), cost_history, marker='o')
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost vs Iteration")
plt.grid(True)
plt.show()
```
